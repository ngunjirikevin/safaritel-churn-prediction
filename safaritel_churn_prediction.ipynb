#### 1.2. Synthetic Dataset Generation

Now we generate the synthetic data of 1000 customers using 5 features. It will simulate a telecommunications customer base

```python
np.random.seed(42) # Set the random seed for reproducibility, bcoz normally running a random number generation prooduces different results each time, so we set a seed to get the same results every time we run the code.
n = 1000

df = pd.DataFrame({
    "monthly_spend_kes": np.random.normal(1500,400,n).clip(100,None), # generates monthly spend data with a normal distribution, mean of 1500, std dev of 400, and clips values below 100 to 100(ensures spending doesnt fall below 100, with no upperlimit imposed).
    "months_active": np.random.randint(1, 60, n), # generates random integers between 1 and 59 inclusive
    "customer_care_calls": np.random.poisson(2, n), # generates random integers from a Poisson distribution with a mean of 2
    "data_bundle_usage_mb": np.random.normal(3000, 900, n).clip(0, None), # generates data bundle usage data with a normal distribution, mean of 3000, std dev of 900, and clips values below 0 to 0
    "late_payments": np.random.poisson(0.7, n), # generates random integers from a Poisson distribution with a mean of 0.7
})

df.head()
```
